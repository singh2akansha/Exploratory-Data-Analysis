```python
plt.figure(figsize = (5,6))
sns.countplot(df['diagnosis'], label = 'count', palette = 'viridis')
plt.show()
```

```python
#one_hot_encoding
def generating_encoding(df,column):
    one_hot = pd.get_dummies(df[column],prefix=column, dtype = int)
    df_encoded = df.join(one_hot)
    df_encoded = df_encoded.drop(column,axis =1)
    return df_encoded

res = generating_encoding(df,'diagnosis')
print(res)
```

#### Coreelation Matrix

```python
corr_df = res.corr()
print(corr_df)
```

```python
corr_df = res.corr()

plt.figure(figsize = (20,18))
sns.heatmap(corr_df, annot= True, square = True, fmt = '.2f',cmap = 'coolwarm')
plt.show()
```

```python
# id,fractal_dimension_se, symmetry_se,smoothness_se, texture_se, fractal_dimension_mean,smoothness_mean,texture_mean,symmetry_mean,compactness_se,concavity_se
res = res.drop(columns = ['id','fractal_dimension_worst','symmetry_worst','texture_worst','radius_se','perimeter_se','area_se','concave points_se','smoothness_worst','fractal_dimension_se', 'symmetry_se','smoothness_se', 'texture_se', 'fractal_dimension_mean','smoothness_mean','texture_mean','symmetry_mean','compactness_se','concavity_se'],axis = 1)
print(res.head())
```

## Bivariate Analysis

```python
# concave points_worst, concavity_worst, area_worst, area_mean
graph = sns.JointGrid(data =df, x ='concave points_worst', y ='area_worst', hue = 'diagnosis' )
graph.plot_joint(sns.scatterplot, alpha = 0.5)
graph.plot_marginals(sns.kdeplot, fill = True)
plt.suptitle('Joint Distribution: Malignant vs Benign  - Area_Worst vs Concave points_worst', y=1.02)
plt.show()
```

```python
graphs = sns.JointGrid(data = df, x = 'concavity_worst', y ='area_worst', hue = 'diagnosis')
graphs.plot_joint(sns.scatterplot, alpha = 0.5)
graphs.plot_marginals(sns.kdeplot, fill = True)
plt.title('Joint Distribution: Malignant vs Benign - Area_Worst vs Concavity_Worst', y = 1.2)
plt.show()
```
